```python
import os
import pandas as pd

# 1. Load Artifact from Notebook 2
input_artifact = "artifacts/02_labeled_orders.csv"
print(f"Loading artifact from: {input_artifact}")
df = pd.read_csv(input_artifact)

# 2. Fast datetime conversion and sorting
print("Parsing dates and sorting...")
df['order_purchase_timestamp'] = pd.to_datetime(df['order_purchase_timestamp'], format='mixed')
df = df.sort_values('order_purchase_timestamp').reset_index(drop=True)

print(f"Total dataset date range: {df['order_purchase_timestamp'].min()} to {df['order_purchase_timestamp'].max()}")
print(f"Total rows: {len(df)}")

# 3. Calculate Split Indices (80% Train, 10% Val, 10% Test)
n_total = len(df)
train_end = int(n_total * 0.80)
val_end = int(n_total * 0.90)

df_train = df.iloc[:train_end].copy()
df_val = df.iloc[train_end:val_end].copy()
df_test = df.iloc[val_end:].copy()

# 4. Display Split Distributions & Target Balance
splits = {
    "Train": df_train,
    "Validation": df_val,
    "Test": df_test
}

print("\n--- Split Summary ---")
for name, data in splits.items():
    date_min = str(data['order_purchase_timestamp'].min())[:10]
    date_max = str(data['order_purchase_timestamp'].max())[:10]
    late_ratio = (data['is_late'].mean()) * 100
    print(f"{name:10} | Rows: {len(data):6} ({len(data)/n_total*100:4.1f}%) | "
          f"Date: {date_min} to {date_max} | Late Rate: {late_ratio:.2f}%")

os.makedirs("artifacts", exist_ok=True)
df_train.to_csv("artifacts/03_train.csv", index=False)
df_val.to_csv("artifacts/03_val.csv", index=False)
df_test.to_csv("artifacts/03_test.csv", index=False)

print("\nArtifacts successfully saved:")
print(" - artifacts/03_train.csv")
print(" - artifacts/03_val.csv")
print(" - artifacts/03_test.csv")
```
